```python
### Using cross-validation and hyperparameter tuning with severity_class as y-target using training set 
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.utils import class_weight
from sklearn.model_selection import train_test_split
from sklearn.metrics import balanced_accuracy_score
```

```python
df=pd.read_csv('final_ohc_wto_sde_train.csv')
df
```

```python
x = df.drop(columns=['severity_class','ev_id'])
y = df[['severity_class']]
```

```python
from sklearn.model_selection import KFold, GridSearchCV
```

```python
k = 5
fold_size = len(x) // k
for i in range(k):
    x_test2 = x.iloc[i * fold_size : (i + 1) * fold_size]
    y_test2= y.iloc[i * fold_size : (i + 1) * fold_size]
    x_train2 = pd.concat([x.iloc[:i * fold_size], x.iloc[(i + 1) * fold_size:]])
    y_train2 = pd.concat([y.iloc[:i * fold_size], y.iloc[(i + 1) * fold_size:]])
```

```python
cv = KFold(n_splits = 5, shuffle=False)  
logReg = LogisticRegression() 
params = {
    'penalty':['l1','l2'],
    'solver': ['liblinear'],
    'C': [0.1, 1, 10, 100],
}

grid_search1 = GridSearchCV(
    estimator=logReg,       
    param_grid=params,      
    scoring = 'balanced_accuracy',
    cv=cv,
    verbose=2,              
    n_jobs=4,              
)

grid_search1.fit(x_train2, y_train2)
```

```python
best_params = grid_search1.best_params_
print("Best Hyperparameters:", best_params)
```

```python
logReg = LogisticRegression(
    penalty='l1',       
    solver='liblinear', 
    C = 100,            
)

logReg.fit(x_train2, y_train2)
y1_pred2 = logReg.predict(x_test2)

balance = balanced_accuracy_score(y_test2, y1_pred2)
print(balance*100)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
accuracy = accuracy_score(y_test2, y1_pred2)
precision = precision_score(y_test2, y1_pred2, average='weighted')  
recall = recall_score(y_test2, y1_pred2, average='weighted')        
f1 = f1_score(y_test2, y1_pred2, average='weighted')
balance = balanced_accuracy_score(y_test2, y1_pred2)
print(balance*100)
print(accuracy *100)
print(precision*100)
print(recall*100)
print(f1*100)
```

```python
dh=pd.read_csv('final_ohc_wto_sde_holdout.csv')
```

```python
x_h = dh.drop(columns=["severity_class","ev_id"]) 
y_h = dh["severity_class"]
```

```python
y_h_pred = logReg.predict(x_h)
```

```python
balance = balanced_accuracy_score(y_h, y_h_pred)
precision = precision_score(y_h, y_h_pred, average='weighted')  
recall = recall_score(y_h, y_h_pred, average='weighted')        
f1 = f1_score(y_h, y_h_pred, average='weighted')

print(balance*100)
print(accuracy *100)
print(precision*100)
print(recall*100)
print(f1*100)
```

```python
### Using cross-validation and hyperparameter tuning with ev_highest_injury as y-target using training set 
```

```python
df=pd.read_csv("final_ohc_wto_sde_train.csv")
df
```

```python
x=df.drop(columns=['ev_highest_injury'])
y=df[['ev_highest_injury']]
```

```python
from sklearn.model_selection import KFold, GridSearchCV
```

```python
k = 5
fold_size = len(x) // k
for i in range(k):
    x_test2 = x.iloc[i * fold_size : (i + 1) * fold_size]
    y_test2= y.iloc[i * fold_size : (i + 1) * fold_size]
    x_train2 = pd.concat([x.iloc[:i * fold_size], x.iloc[(i + 1) * fold_size:]])
    y_train2 = pd.concat([y.iloc[:i * fold_size], y.iloc[(i + 1) * fold_size:]])
```

```python
x_train2 = x_train2.drop(columns=['ev_id'], errors='ignore') 
x_test2 = x_test2.drop(columns=['ev_id'], errors='ignore')
```

```python
logReg = LogisticRegression(
    penalty='l1',       
    solver='liblinear', 
    C = 10,            
)

logReg.fit(x_train2, y_train2)
y1_pred2 = logReg.predict(x_test2)

balance = balanced_accuracy_score(y_test2, y1_pred2)
print(balance*100)
```

```python
cv = KFold(n_splits = 5, shuffle=False)  
logReg = LogisticRegression() 
params = {
    'penalty':['l1','l2'],
    'solver': ['liblinear'],
    'C': [0.1, 1, 10, 100],
}

grid_search3 = GridSearchCV(
    estimator=logReg,       
    param_grid=params,      
    scoring = 'balanced_accuracy',
    cv=cv,
    verbose=2,              
    n_jobs=4,              
)

grid_search3.fit(x_train2, y_train2)
```

```python
y1_pred2 = grid_search3.predict(x_test2)
accuracy = accuracy_score(y_test2, y1_pred2)
precision = precision_score(y_test2, y1_pred2, average='weighted')  
recall = recall_score(y_test2, y1_pred2, average='weighted')        
f1 = f1_score(y_test2, y1_pred2, average='weighted')
balance = balanced_accuracy_score(y_test2, y1_pred2)
print(balance*100)
print(accuracy *100)
print(precision*100)
print(recall*100)
print(f1*100)
```

```python
joblib.dump(grid_search3, "LR_grid_search3.pkl")
```
